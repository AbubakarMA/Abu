```python
import numpy as np
```

```python
#PCA from scratch using eigendecomposition
np.random.seed(42)

#Generate the data in 2D with covariance

mean = [0, 0]
#the correlated features
cov = [[3, 1.5], [1.5, 1]]

data = np.random.multivariate_normal(mean, cov, 100)


#step 1 is to center the data by substracting the mean

data_centered = data - data.mean(axis=0)

#Step  2 is to compute the variance matrix
cov_matrix = np.cov(data_centered.T)
print(f"Covariance matrix:\n{cov_matrix}")

#Step 3 is to compute the eigendecomposition of the covariance matrix
eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

print(f"\nEigenvalues: {eigenvalues}")
print(f"\nEigenvectors: {eigenvectors}")

#Eigen vector with the largest eigenvalue is the first principal component!
#And points in the direction of the maximum variance

idx = eigenvalues.argsort()[::-1]
principal_direction = eigenvectors[:, idx]

print(f"\nFirst Principal Component Direction:{principal_direction}")

#step 4 is to project the data onto the principal components

projected = data_centered @ eigenvectors[:, idx]

print(f"\nVariance Explained by each component:")

total_var = sum(eigenvalues)
for i, ev in enumerate(sorted(eigenvalues, reverse=True)):
  print(f" PC{i+1}: {ev/total_var*100:.1f}%")

#Comparing with sklearn PCA

from sklearn.decomposition import PCA
pca = PCA(n_components=2)
pca.fit(data_centered)
print(f"\nSklearn Explained variance ratio: {pca.explained_variance_ratio_}")
print(f"Sklearn components: \n{pca.components_}")
```
