```python
import numpy as np

def sigmoid(z):
    """Squashes any real number into the (0, 1) range."""
    # We clip z to prevent overflow warnings in np.exp()
    z = np.clip(z, -250, 250)
    return 1 / (1 + np.exp(-z))

def compute_loss(y_true, y_pred):
    """Binary Cross-Entropy Loss."""
    m = len(y_true)
    # Add a tiny epsilon (1e-15) to prevent log(0) which causes NaN errors
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)

    loss = - (1/m) * np.sum(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))
    return loss

def train_logistic_regression(X, y, learning_rate=0.1, epochs=1000):
    """Trains the model using Gradient Descent."""
    m, n = X.shape
    w = np.zeros(n)
    b = 0.0

    for epoch in range(epochs):
        # 1. Forward Pass (Linear combination + Sigmoid)
        z = np.dot(X, w) + b
        y_pred = sigmoid(z)

        # 2. Calculate Gradients (The Calculus Magic we derived!)
        error = y_pred - y
        dw = (1/m) * np.dot(X.T, error)
        db = (1/m) * np.sum(error)

        # 3. Update Weights
        w -= learning_rate * dw
        b -= learning_rate * db

        if epoch % 200 == 0:
            print(f"Epoch {epoch} | Loss: {compute_loss(y, y_pred):.4f}")

    return w, b

# --- Test it on fake data ---
np.random.seed(42)
X_train = np.random.randn(100, 2) # 100 samples, 2 features
# Create a linear boundary: if x1 + x2 > 0, then class 1, else class 0
y_train = (X_train[:, 0] + X_train[:, 1] > 0).astype(int)

w_learned, b_learned = train_logistic_regression(X_train, y_train)
print(f"\nLearned Weights: {w_learned}, Bias: {b_learned:.4f}")
```
