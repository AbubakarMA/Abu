A. From Scratch (Pure Python)


```python
def calculate_metrics(y_true, y_pred):
    """Calculates Precision, Recall, and F1 from scratch."""
    tp = sum(1 for yt, yp in zip(y_true, y_pred) if yt == 1 and yp == 1)
    fp = sum(1 for yt, yp in zip(y_true, y_pred) if yt == 0 and yp == 1)
    fn = sum(1 for yt, yp in zip(y_true, y_pred) if yt == 1 and yp == 0)

    # Guard against division by zero
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0.0

    # Harmonic Mean (F1-Score)
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0

    return precision, recall, f1

# Test Case: A model that catches 3 frauds, misses 2, and has 1 false alarm
y_actual = [1, 1, 1, 1, 1, 0, 0, 0] # 5 Frauds, 3 Legits
y_model  = [1, 1, 1, 0, 0, 1, 0, 0] # Predicted 4 Frauds

p, r, f1 = calculate_metrics(y_actual, y_model)
print(f"Precision: {p:.2f} (3 out of 4 alarms were real)") # 3 / 4 = 0.75
print(f"Recall:    {r:.2f} (3 out of 5 actual frauds caught)") # 3 / 5 = 0.60
print(f"F1-Score:  {f1:.2f} (Harmonic Mean)") # ~0.66
```

B.The Production Library (Scikit-Learn)

```python
from sklearn.metrics import precision_score, recall_score, f1_score, roc_auc_score

# Scikit-learn handles the edge cases and zero-divisions automatically
print(f"SKLearn Precision: {precision_score(y_actual, y_model):.2f}")
print(f"SKLearn Recall:    {recall_score(y_actual, y_model):.2f}")
print(f"SKLearn F1:        {f1_score(y_actual, y_model):.2f}")

# For AUC, we need the raw probabilities, not the 0/1 predictions!
y_probs = [0.9, 0.8, 0.7, 0.4, 0.3, 0.6, 0.1, 0.2]
print(f"SKLearn AUC:       {roc_auc_score(y_actual, y_probs):.2f}")
```
